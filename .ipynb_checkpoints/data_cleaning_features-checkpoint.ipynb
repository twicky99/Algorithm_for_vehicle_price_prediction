```python
# Lets remove 29 brands that are not popular out of 65
remove = ['Jeep','Lexus','Mini','Range-Rover','Skoda','Foton','Chrysler','JMC','Jonway','Ashok-Leyland',
 'Opel','Ssangyong','Acura','FAW','Force','Citroen','Sakai','Atco','Changan','Eicher','JiaLing',
 'Seat','Hummer','Porsche','Alfa-Romeo','Hero-Honda','Wave','Cal','Vauxhall']
```

```python
len(remove)
```

```python
df = df[(df['make'] != remove[0]) & (df['make'] != remove[1]) & (df['make'] != remove[2]) &
        (df['make'] != remove[3]) & (df['make'] != remove[4]) & (df['make'] != remove[5]) &
        (df['make'] != remove[6]) & (df['make'] != remove[7]) & (df['make'] != remove[8]) &
        (df['make'] != remove[9]) & (df['make'] != remove[10]) & (df['make'] != remove[11]) &
        (df['make'] != remove[12]) & (df['make'] != remove[13]) & (df['make'] != remove[14]) &
        (df['make'] != remove[15]) & (df['make'] != remove[16]) & (df['make'] != remove[17]) &
        (df['make'] != remove[18]) & (df['make'] != remove[19]) & (df['make'] != remove[20]) &
        (df['make'] != remove[21]) & (df['make'] != remove[22]) & (df['make'] != remove[23]) &
        (df['make'] != remove[24]) & (df['make'] != remove[25]) & (df['make'] != remove[26]) &
        (df['make'] != remove[27]) & (df['make'] != remove[28])]
```

```python
df['make'].nunique()
```

```python
df['make'].unique()
```

```python
df.shape
```

- There is an 'Other' brand in Make.
- Let's drop it as well.

```python
df = df[df['make'] != 'Other']
```

```python
df.shape
```

```python
df.info()
```

- We are done feature engineering the model
- We need to create dummy variables for all the unique make items which we will do at the end.

```python
df.to_csv('data\\vehicle_dataset_after_model.csv', encoding='utf-8', index=False)
```
