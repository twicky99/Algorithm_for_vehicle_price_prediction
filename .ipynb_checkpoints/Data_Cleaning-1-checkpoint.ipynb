- It seems that all different kind of categories are having price as 'Negotiable' and it's not applicable to a single category.

- However, filling the 'Negotiable' with a price is an option that is tedious and time consuming. Hence Let's drop them.

```python
df[df['price']=='Negotiable']['model'].value_counts().head(10) 
```

- There is no clear pattern in the price 'Negotiable'.
- It seems putting the price as 'Negotiable' is common among the small vehicles like Alto, Virtz, Every, Maruti..

```python
# Removing the price field where the value is 'Negotiable'
df = df[df['price']!='Negotiable']
```

```python
# Check if it's removed
df.shape
```

```python
# Check if price 'Negotiable' still exists
price_negotiable = len(df['price'].loc[df['price'].str.startswith('N', na=False)])
price_negotiable
```

 - We have successfully removed the 'Negotiable' values.
 - Let's remove the Rs. field from price column
 - Analyse the price field to see if anything odd appears

```python
# spliting the numeric value from the price column
# Testing a sample
df['price'][0][4:]
```

```python
# making sure that all the values are starting with 'Rs. '
price_with_Rs = len(df['price'].loc[df['price'].str.startswith('Rs. ', na=False)])
price_with_Rs
```

```python
# removing the Rs. field and replacing with the existing value
# ignore the warning as python is worried that we are replacing the data
df['price'] = df['price'].apply(lambda x:x[4:])
```

```python
df.head()
```

```python
df.info()
```

- Since the price column is not numeric, we need to convert it to numeric
- One issue we have is that the prices are with commas (,) which prevents us from converting to numeric.
- We need to remove the commas (,) and then convert to numerics

```python
# Remove the commas (,)
df['price'] = df['price'].apply(lambda x: x.replace(',', ''))
```

```python
# Convert the price column to numeric
df['price'] = pd.to_numeric(df['price'])
```

```python
df.info()
```

- It's not possible to have values less than 600,000. 
- There are 452 records shows less than 600,000.
- This either due to missing a 0 at the end or the vehicle is priced less due to faulty.
- Dropping these rows will definitely reduce the data but it's relatively small
- To begin with, let's drop the rows with those values.

```python
# Removing the values less than 600,000
df = df[df['price'] > 600000]
```

```python
df.head()
```

```python
# Double check if the value less than 600,000 are dropped
len(df[df['price'] < 600000])
```

```python
df[df['price'] < 900000]['model'].value_counts()
```

 - As per the data, Nano, QQ and Maruti can have prices from 600k.
 - Other vehicle are generally over 900k.
 - Hence, we want to keep the data for Nano, Maruti and QQ between 600k and 900k while dropping all other data below 900k
 - Let's create some dataframes for nano, QQ and Maruti

```python
# Creating Dataframe for nano, maruti and QQ vehicles that are between 600k and 900k
df_nano = df[(df['price'] < 900000) & (df['model'] == 'Nano')]
df_qq = df[(df['price'] < 900000) & (df['model'] == 'QQ')]
df_Maruti = df[(df['price'] < 900000) & (df['model'] == 'Maruti')]
```

```python
print(df_nano.shape[0])
print(df_qq.shape[0])
print(df_Maruti.shape[0])
```

```python
# From the original dataframe, extract only data that are over 900k
df_over_900k = df[df['price'] > 900000]
```

```python
df_lists = [df_over_900k, df_nano,df_qq, df_Maruti]
```

```python
# Combine the data that is over 900k and Nano, QQ and Maruti that are between 600k and 900k
df = pd.concat(df_lists)
```
